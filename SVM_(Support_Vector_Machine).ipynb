```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
dataset = pd.read_csv('iris-flower.csv')
dataset
```

```python
dataset['Species'].unique()
```

```python
<StringArray>
['Iris-sentosa', 'Iris-versicolor', 'Iris-virginica']
Length: 3, dtype: str
```

```python
le = LabelEncoder()
dataset['Species_encoded'] = le.fit_transform(dataset['Species'])

x = dataset.drop(colums=['Id', 'Species', 'Species_encoded'], errors = 'ignore')
y = dataset['Species_encoded']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)
```

```python
pipeline = Pipeline([('scaler', StandardScaler()),('svc', SVC())])
```

```python
params = {
    'svc_kernel' : ['poly', 'rbf', 'sigmoid'],
    'svc_C' : [0.5, 1, 10, 100],
    'svc_gamma' : ['scale', 1, 0.1, 0.001]
}
```

```python
grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=params,
    cv=5,
    scoring='accuracy'
    n_jobs = 1,
    verbose = 1
)
```
