```python
# 202431049_TriPrasetya
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 202431049_TriPrasetya
img = cv2.imread("foto.jpeg")

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# 202431049_TriPrasetya
(h, w) = img.shape[:2]
center = (w // 2, h // 2)

M = cv2.getRotationMatrix2D(center, 45, 1)

rotated = cv2.warpAffine(img, M, (w, h))
```

```python
# 202431049_TriPrasetya
resized = cv2.resize(img, (300, 400))
```

```python
# 202431049_TriPrasetya
cropped = img[100:700, 250:850]
```

```python
# 202431049_TriPrasetya
flipped = cv2.flip(img, 1)
```

```python
# 202431049_TriPrasetya
M = np.float32([[1, 0, 80],
                [0, 1, 60]])

translated = cv2.warpAffine(img, M, (w, h))
```

```python
# 202431049_TriPrasetya
plt.figure(figsize=(15,10))

plt.subplot(2,3,1)
plt.imshow(img)
plt.title("Citra Asli")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(rotated)
plt.title("Rotated")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(resized)
plt.title("Resized")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(cropped)
plt.title("Cropped")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(flipped)
plt.title("Flipped")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(translated)
plt.title("Translated")
plt.axis("off")

plt.tight_layout()
plt.show()
```
