```python
import requests
import pandas as pd
from io import StringIO

header = {"User-Agent": "Mozilla/5.0"}

#WEB SCRAPE 1
url_worldmeter = "https://www.worldometers.info/world-population/population-by-country/"

req_worldmeter = requests.get(
    url_worldmeter, headers=header)

html = req_worldmeter.content.decode("utf-8")

tabel_worldmater = pd.read_html(StringIO(html))
df_populasi = tabel_worldmater[0]

print("Data Scraping Worldmeter:")
display(df_populasi.head())
```

```python
#WEB SCRAPE 2
url_scrape = "https://www.scrapethissite.com/pages/forms/"
req_scrape = requests.get(url_scrape, headers=header)

tabel_scrape = pd.read_html(StringIO(req_scrape.text))

df_hoki = tabel_scrape[0]

print("Data Scraping ScrapeThisSite:")
display(df_hoki.head())
```
